```python
import pandas as pd


df_model = pd.read_csv(
    "../data/processed/final_features_v1.csv"
)


df_model.shape
```

```python
X = df_model.drop(
    columns=["SeriousDlqin2yrs"]
)

y = df_model["SeriousDlqin2yrs"]


print(X.shape)
print(y.value_counts())
```

```python
from sklearn.model_selection import StratifiedKFold


skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)


def evaluate_model(
    model,
    X_test,
    y_test
):

    probabilities = model.predict_proba(
        X_test
    )[:,1]


    predictions = (
        probabilities >= 0.5
    ).astype(int)


    return {
        "ROC-AUC":
            roc_auc_score(
                y_test,
                probabilities
            ),

        "PR-AUC":
            average_precision_score(
                y_test,
                probabilities
            ),

        "Precision":
            precision_score(
                y_test,
                predictions
            ),

        "Recall":
            recall_score(
                y_test,
                predictions
            ),

        "F1":
            f1_score(
                y_test,
                predictions
            ),

        "Confusion Matrix":
            confusion_matrix(
                y_test,
                predictions
            )
    }
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression


logistic_model = Pipeline(
    steps=[
        (
            "scaler",
            StandardScaler()
        ),
        (
            "model",
            LogisticRegression(
                max_iter=1000,
                class_weight="balanced",
                random_state=42
            )
        )
    ]
)
```

```python
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)


logistic_probs = cross_val_predict(
    logistic_model,
    X,
    y,
    cv=skf,
    method="predict_proba",
    n_jobs=-1
)[:,1]


logistic_preds = (
    logistic_probs >= 0.5
).astype(int)
```

```python
logistic_results = {
    "ROC-AUC":
        roc_auc_score(
            y,
            logistic_probs
        ),

    "PR-AUC":
        average_precision_score(
            y,
            logistic_probs
        ),

    "Precision":
        precision_score(
            y,
            logistic_preds
        ),

    "Recall":
        recall_score(
            y,
            logistic_preds
        ),

    "F1":
        f1_score(
            y,
            logistic_preds
        ),

    "Confusion Matrix":
        confusion_matrix(
            y,
            logistic_preds
        )
}


logistic_results
```

```python
from sklearn.ensemble import RandomForestClassifier


rf_model = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)
```

```python
rf_probs = cross_val_predict(
    rf_model,
    X,
    y,
    cv=skf,
    method="predict_proba",
    n_jobs=-1
)[:,1]


rf_preds = (
    rf_probs >= 0.5
).astype(int)
```

```python
rf_results = {
    "ROC-AUC":
        roc_auc_score(
            y,
            rf_probs
        ),

    "PR-AUC":
        average_precision_score(
            y,
            rf_probs
        ),

    "Precision":
        precision_score(
            y,
            rf_preds
        ),

    "Recall":
        recall_score(
            y,
            rf_preds
        ),

    "F1":
        f1_score(
            y,
            rf_preds
        ),

    "Confusion Matrix":
        confusion_matrix(
            y,
            rf_preds
        )
}


rf_results
```

```python
from xgboost import XGBClassifier


class_ratio = (
    y.value_counts()[0] /
    y.value_counts()[1]
)


xgb_model = XGBClassifier(
    n_estimators=300,
    learning_rate=0.05,
    scale_pos_weight=class_ratio,
    random_state=42,
    eval_metric="logloss",
    n_jobs=-1
)
```

```python
xgb_probs = cross_val_predict(
    xgb_model,
    X,
    y,
    cv=skf,
    method="predict_proba",
    n_jobs=-1
)[:,1]


xgb_preds = (
    xgb_probs >= 0.5
).astype(int)
```

```python
xgb_results = {
    "ROC-AUC":
        roc_auc_score(
            y,
            xgb_probs
        ),

    "PR-AUC":
        average_precision_score(
            y,
            xgb_probs
        ),

    "Precision":
        precision_score(
            y,
            xgb_preds
        ),

    "Recall":
        recall_score(
            y,
            xgb_preds
        ),

    "F1":
        f1_score(
            y,
            xgb_preds
        ),

    "Confusion Matrix":
        confusion_matrix(
            y,
            xgb_preds
        )
}


xgb_results
```

```python
from lightgbm import LGBMClassifier


class_ratio = (
    y.value_counts()[0] /
    y.value_counts()[1]
)


lgbm_model = LGBMClassifier(
    n_estimators=300,
    learning_rate=0.05,
    scale_pos_weight=class_ratio,
    random_state=42,
    n_jobs=-1
)
```

```python
lgbm_results = {
    "ROC-AUC":
        roc_auc_score(
            y,
            lgbm_probs
        ),

    "PR-AUC":
        average_precision_score(
            y,
            lgbm_probs
        ),

    "Precision":
        precision_score(
            y,
            lgbm_preds
        ),

    "Recall":
        recall_score(
            y,
            lgbm_preds
        ),

    "F1":
        f1_score(
            y,
            lgbm_preds
        ),

    "Confusion Matrix":
        confusion_matrix(
            y,
            lgbm_preds
        )
}


lgbm_results
```

```python
from catboost import CatBoostClassifier


cat_model = CatBoostClassifier(
    iterations=300,
    learning_rate=0.05,
    random_state=42,
    verbose=0,
    thread_count=-1
)
```

```python
import numpy as np


sample_weights = np.where(
    y == 1,
    class_ratio,
    1
)
```

```python
from sklearn.base import clone


cat_probs = np.zeros(len(y))


for train_idx, valid_idx in skf.split(X, y):

    X_train_fold = X.iloc[train_idx]
    X_valid_fold = X.iloc[valid_idx]

    y_train_fold = y.iloc[train_idx]
    
    weights_fold = sample_weights[train_idx]


    model = clone(cat_model)


    model.fit(
        X_train_fold,
        y_train_fold,
        sample_weight=weights_fold
    )


    cat_probs[valid_idx] = model.predict_proba(
        X_valid_fold
    )[:,1]
```

```python
cat_preds = (
    cat_probs >= 0.5
).astype(int)
```

```python
cat_results = {
    "ROC-AUC":
        roc_auc_score(
            y,
            cat_probs
        ),

    "PR-AUC":
        average_precision_score(
            y,
            cat_probs
        ),

    "Precision":
        precision_score(
            y,
            cat_preds
        ),

    "Recall":
        recall_score(
            y,
            cat_preds
        ),

    "F1":
        f1_score(
            y,
            cat_preds
        ),

    "Confusion Matrix":
        confusion_matrix(
            y,
            cat_preds
        )
}


cat_results
```

```python
baseline_summary = baseline_summary.sort_values(
    by=[
        "ROC-AUC",
        "PR-AUC",
        "Recall"
    ],
    ascending=False
)


baseline_summary.insert(
    0,
    "Rank",
    range(1, len(baseline_summary)+1)
)


baseline_summary
```
