```python
# ============================================================
# IMPORTS
# ============================================================

import yaml
import joblib
import numpy as np
import pandas as pd

from catboost import CatBoostClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    precision_score,
    recall_score,
    f1_score,
    accuracy_score
)
```

```python
# ============================================================
# LOAD DATA
# ============================================================

TARGET = "SeriousDlqin2yrs"

df = pd.read_csv(
    "../data/processed/final_modeling_dataset_v1.csv"
)

X = df.drop(columns=TARGET)
y = df[TARGET]
```

```python
# ============================================================
# TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    stratify=y,

    random_state=42

)
```

```python
# ============================================================
# LOAD PARAMETERS
# ============================================================

with open(
    "../config/params_all.yaml",
    "r"
) as f:

    params = yaml.safe_load(f)
```

```python
# ============================================================
# SCALE POS WEIGHT
# ============================================================

scale_pos_weight = (

    (y_train == 0).sum()

    /

    (y_train == 1).sum()

)
```

```python
# ============================================================
# FINAL CATBOOST
# ============================================================

cat_params = params["models"]["catboost"]["params"]

cat_fixed = params["models"]["catboost"]["fixed_params"]

cat_model = CatBoostClassifier(

    **cat_params,

    **cat_fixed

)

cat_model.fit(

    X_train,
    y_train

)
```

```python
# ============================================================
# FINAL XGBOOST
# ============================================================

xgb_params = params["models"]["xgboost"]["params"]

xgb_fixed = params["models"]["xgboost"]["fixed_params"]

xgb_fixed["scale_pos_weight"] = scale_pos_weight

xgb_model = XGBClassifier(

    **xgb_params,

    **xgb_fixed

)

xgb_model.fit(

    X_train,
    y_train

)
```

```python
# ============================================================
# FINAL LIGHTGBM
# ============================================================

lgb_params = params["models"]["lightgbm"]["params"]

lgb_fixed = params["models"]["lightgbm"]["fixed_params"]

lgb_model = LGBMClassifier(

    **lgb_params,

    **lgb_fixed

)

lgb_model.fit(

    X_train,
    y_train

)
```

```python
# ============================================================
# PROBABILITIES
# ============================================================

cat_probs = cat_model.predict_proba(
    X_test
)[:,1]

xgb_probs = xgb_model.predict_proba(
    X_test
)[:,1]

lgb_probs = lgb_model.predict_proba(
    X_test
)[:,1]
```

```python
# ============================================================
# THRESHOLD OPTIMIZATION
# ============================================================

def optimize_threshold(

    y_true,

    probs,

    recall_target=0.75

):

    rows=[]

    for threshold in np.arange(

        0.01,
        1.00,
        0.01

    ):

        preds = (

            probs >= threshold

        ).astype(int)

        precision = precision_score(
            y_true,
            preds,
            zero_division=0
        )

        recall = recall_score(
            y_true,
            preds
        )

        f1 = f1_score(
            y_true,
            preds
        )

        rows.append({

            "Threshold":threshold,

            "Precision":precision,

            "Recall":recall,

            "F1":f1

        })

    df = pd.DataFrame(rows)

    df = df[
        df["Recall"] >= recall_target
    ]

    return df.sort_values(

        "F1",

        ascending=False

    ).reset_index(drop=True)
```

```python
# ============================================================
# RUN THRESHOLD SEARCH
# ============================================================

cat_thresholds = optimize_threshold(

    y_test,

    cat_probs

)

xgb_thresholds = optimize_threshold(

    y_test,

    xgb_probs

)

lgb_thresholds = optimize_threshold(

    y_test,

    lgb_probs

)
```

```python
# ============================================================
# BEST THRESHOLDS
# ============================================================

display(cat_thresholds.head())

display(xgb_thresholds.head())

display(lgb_thresholds.head())
```

# Ensamble Experiments

```python
# ============================================================
# ENSEMBLE PROBABILITIES
# ============================================================

ensemble_probs = {

    "Equal Voting":
        (
            cat_probs +
            xgb_probs +
            lgb_probs
        ) / 3,

    "Weighted 50-25-25":
        (
            0.50 * cat_probs +
            0.25 * xgb_probs +
            0.25 * lgb_probs
        ),

    "Weighted 40-30-30":
        (
            0.40 * cat_probs +
            0.30 * xgb_probs +
            0.30 * lgb_probs
        ),

    "CatBoost + XGBoost":
        (
            cat_probs +
            xgb_probs
        ) / 2,

    "CatBoost + LightGBM":
        (
            cat_probs +
            lgb_probs
        ) / 2,

    "XGBoost + LightGBM":
        (
            xgb_probs +
            lgb_probs
        ) / 2

}
```

```python
# ============================================================
# THRESHOLD OPTIMIZATION
# ============================================================

ensemble_results = {}

for name, probs in ensemble_probs.items():

    print("=" * 80)
    print(name)
    print("=" * 80)

    result = optimize_threshold(
        y_test,
        probs,
        recall_target=0.75
    )

    ensemble_results[name] = result

    display(
        result.head()
    )
```

```python
# ============================================================
# FINAL ENSEMBLE EVALUATION
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    average_precision_score,
    precision_score,
    recall_score,
    f1_score
)

comparison = []

for name, probs in ensemble_probs.items():

    best = ensemble_results[name].iloc[0]

    threshold = best["Threshold"]

    preds = (
        probs >= threshold
    ).astype(int)

    comparison.append({

        "Model": name,

        "Threshold": threshold,

        "Accuracy":
            accuracy_score(
                y_test,
                preds
            ),

        "ROC-AUC":
            roc_auc_score(
                y_test,
                probs
            ),

        "PR-AUC":
            average_precision_score(
                y_test,
                probs
            ),

        "Precision":
            precision_score(
                y_test,
                preds
            ),

        "Recall":
            recall_score(
                y_test,
                preds
            ),

        "F1":
            f1_score(
                y_test,
                preds
            )

    })

comparison_df = (
    pd.DataFrame(comparison)
    .sort_values(
        by="F1",
        ascending=False
    )
    .reset_index(drop=True)
)

display(comparison_df)
```
